C'è un laser a diodo che presumibilmente emette un fasio leggermente divergente, si mette quindi una lente collimatrice che rende tale fascio parallelo.

Si procede quindi con l'acquisizione delle curve IV. Si parte con una misura della corrente di buio (fatta principalmente dalla torcia del telefono, necessaria per vedere lo schermo LCD del misuratore di potenza). La potenza di buio al buio è $$P_{buio} = 1.15\, \text{nW}$$
Potenza di buio a luce accesa $$P = 5.3\, \mu\text{W}$$ che è 3 ordini di grandezza sotto la potenza che si misura a 80mA ma circa 1/3 di quella a 20mA. Dunque l'acqusizione è da fare a luce spenta, con al torcia del telefono ad illuminare LCD.



# Plot IW a temperatura ambiente

```python
import numpy as np
import matplotlib

import matplotlib.pyplot as plt
matplotlib.use('TkAgg')

data = np.loadtxt('/Users/valeriofrancopecchia/Desktop/Fisica/LabMateria/Diodo Laser/IW_T_ambiente.txt', skiprows=2)
I = data[:, 0]  # Current in mA
W = data[:, 1]  # Power in mW

plt.errorbar(I, W, color='blue', label='Experimental Data', fmt='.')
plt.xlabel('Current [mA]')
plt.ylabel('Power [mW]')
plt.title('Laser Diode Power vs Current at Ambient Temperature')
plt.legend()

plt.show(block = True)
```

# Plot di tutte le curve
E' stata aggiunta una nuova acquisizione a 22-23°C. effettivi, in quanto quella che doveva essere a T_amb non abbiamo la certezza fosse a 20°C, ma probabilmente 18 o 17. Sotto sono riportate tutte le curve con le label.

```python
import numpy as np
import matplotlib

import matplotlib.pyplot as plt
matplotlib.use('TkAgg')

data = np.loadtxt('/Users/valeriofrancopecchia/Desktop/Fisica/LabMateria/Diodo Laser/IW_T_13-15.txt', skiprows=2)
I = data[:, 0]  # Current in mA
W = data[:, 1]  # Power in mW
data_a = np.loadtxt('/Users/valeriofrancopecchia/Desktop/Fisica/LabMateria/Diodo Laser/IW_T_ambiente.txt', skiprows=2)
I_a = data_a[:, 0]  # Current in mA
W_a = data_a[:, 1]  # Power in mW
data_c = np.loadtxt('/Users/valeriofrancopecchia/Desktop/Fisica/LabMateria/Diodo Laser/IW_T_43-44.txt', skiprows=2)
I_c = data_c[:, 0]  # Current in mA
W_c = data_c[:, 1]  # Power in mW
data_2 = np.loadtxt('/Users/valeriofrancopecchia/Desktop/Fisica/LabMateria/Diodo Laser/IW_T_22-23.txt', skiprows=2)
I_2 = data_2[:, 0]  # Current in mA
W_2 = data_2[:, 1]  # Power in mW

plt.errorbar(I, W, color='blue', label='T=13-15°C', fmt='.-')
plt.errorbar(I_a, W_a, color='black', label='T=18°C?', fmt='.-')
plt.errorbar(I_c, W_c, color='red', label='T=43-44°C', fmt='.-')
plt.errorbar(I_2, W_2, color='orange', label='T=22-23°C', fmt='.-')
plt.xlabel('Current [mA]')
plt.ylabel('Power [mW]')
plt.title('Laser Diode Power vs Current in function of Temperature')
plt.legend()

plt.show(block = True)
```
